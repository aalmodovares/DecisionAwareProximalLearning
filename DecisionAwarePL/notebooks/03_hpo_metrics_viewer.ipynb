# HPO Metrics Viewer
Carga un `metrics.csv` generado por `scripts/search_hparams.py` y ordena trials
por la metrica que elijas.
El CSV contiene hiperparametros y metricas `val_*` y `test_*` por trial.


```python
# Ajusta esta ruta a tu corrida HPO
from pathlib import Path
import pandas as pd

def load_metrics(path):
    path = Path(path)
    if not path.exists():
        raise FileNotFoundError(f"No existe: {path}")
    frame = pd.read_csv(path)
    print(f"Rows: {len(frame)} | Cols: {len(frame.columns)}")
    display(frame.head(10))
    return frame


def rank_metrics(frame, metric, ascending=True, topk=30):
    if metric not in frame.columns:
        raise KeyError(f"Metrica no encontrada: {metric}.")
    ranked = frame.sort_values(metric, ascending=ascending).reset_index(drop=True)
    display(ranked.head(topk))
    return ranked


metrics_csv = Path("../search_results/semisynthetic_tcga_v1_proximal_unweighted_factual_rmse_optuna/nmmr_u/metrics.csv")
df = load_metrics(metrics_csv)
```

```python
# Seleccion de ordenacion

sort_metric = "val_regret"  # ejemplo: val_factual_mean_rmse, test_policy_value, val_global_rmse

ascending = True # True para RMSE/MAE, False para policy_value
topk = 30

ranked = rank_metrics(df, sort_metric, ascending=ascending, topk=topk)
```

```python
# Inspect a selected metric.
metric = "objective_value" if "objective_value" in df.columns else df.columns[0]
ranked = rank_metrics(df, metric, ascending=True, topk=30)
```
